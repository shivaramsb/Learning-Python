## 1. Helper Function: `hex_id`
Before we start, let's create a helper to see memory addresses in Hexadecimal format (like `0x7ff...`), which is the standard way to view memory locations.

```python
def hex_id(obj):
    return hex(id(obj))

x = 10
print(f"Value: {x}")
print(f"Memory Address: {hex_id(x)}")
```

## 2. Variables as References
In Python, variables are **labels** pointing to objects in memory. They are not boxes containing values.

When you say `a = [1, 2]` and `b = a`, you are NOT copying the list. You are creating a second tag for the **same** list.

```python
# List Alias Example
a = [1, 2, 3]
b = a  # b points to the SAME object as a

print(f"Address of a: {hex_id(a)}")
print(f"Address of b: {hex_id(b)}")

# Modifying b changes a!
b.append(4)
print(f"a after modyfing b: {a}")

# Check identity
print(f"Is a exactly b? {a is b}")
```

## 3. The Difference Between `==` and `is`
- `==` checks **Value Equality** (Do they contain the same data?)
- `is` checks **Reference Equality** (Do they point to the exact same memory address?)

> **Interview Question:** When would `a == b` be True but `a is b` be False?

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(f"Values match? {list1 == list2}")    # True, contents are same
print(f"Identities match? {list1 is list2}") # False, they are two different lists in memory
print(f"Addr 1: {hex_id(list1)}")
print(f"Addr 2: {hex_id(list2)}")
```

## 4. Integer Optimization (Small Integer Caching)
Python pre-allocates small integers (usually -5 to 256) at startup. Every time you use `100`, you are referencing the same global `100` object.

```python
a = 100
b = 100
print(f"100 is 100? {a is b}") # True, cached

x = 300
y = 300
# Note: In Jupyter/PyCharm, sometimes this returns True due to compiler optimization.
# But in a standard REPL or script, distinct large integers are different objects.
print(f"300 is 300? {x is y}")
```

## 5. Arbitrary Precision Integers
Other languages (C, Java) have limits on integers (e.g., 64-bit max value is ~9 quintillion). 
Python integers objects automatically expand to use as much RAM as needed. You can calculate $2^{1000}$ without overflow.

```python
import sys

huge_num = 2 ** 1000
print(f"Huge Number (first 50 digits): {str(huge_num)[:50]}...")

# Check memory usage of this integer
print(f"Size in bytes: {sys.getsizeof(huge_num)}")

# Standard int size
print(f"Size of number 1: {sys.getsizeof(1)}")
```

## 6. String Interning (Optimization)
Python automatically "interns" (caches) short strings that look like identifiers (only letters/numbers/underscores). This saves memory and makes comparisons faster.

```python
s1 = "hello"
s2 = "hello"
print(f"'hello' is 'hello'? {s1 is s2}") # True (Interned)

# Strings with special chars or spaces are usually NOT interned automatically
s3 = "hello world!"
s4 = "hello world!"
print(f"'hello world!' is 'hello world!'? {s3 is s4}") # False (New object created)

# You can force interning
import sys
s5 = sys.intern("hello world!")
s6 = sys.intern("hello world!")
print(f"Forced intern: {s5 is s6}")
```

## 7. Bitwise Operations 🤖
Computers think in binary. Bitwise operators let you manipulate individual bits of an integer.

| Operator | Name | Description |
|---|---|---|
| `&` | AND | Returns 1 if both bits are 1 |
| `|` | OR | Returns 1 if one of the bits is 1 |
| `^` | XOR | Returns 1 if bits are different |
| `~` | NOT | Inverts all bits |
| `<<` | Left Shift | Shifts bits to left (Multiply by 2) |
| `>>` | Right Shift | Shifts bits to right (Divide by 2) |

```python
a = 10  # Binary: 1010
b = 4   # Binary: 0100

print(f"a = {bin(a)}")
print(f"b = {bin(b)}")

print(f"a & b = {a & b} (Binary: {bin(a & b)}) 1010 & 0100 = 0000")
print(f"a | b = {a | b} (Binary: {bin(a | b)}) 1010 | 0100 = 1110")
print(f"a ^ b = {a ^ b} (Binary: {bin(a ^ b)}) 1010 ^ 0100 = 1110")

# Deep Dive Tip: x << n is same as x * (2**n)
x = 5 # 101
print(f"5 << 1 = {5 << 1} (1010)")
print(f"5 << 2 = {5 << 2} (10100)")
```

## 8. Floating Point Precision (IEEE 754)
Never rely on perfect equality for floats. Computers store decimals in binary fractions, which leads to precision errors.

```python
val = 0.1 + 0.2
print(f"0.1 + 0.2 = {val}") # It is NOT 0.3!

print(f"Is it 0.3? {val == 0.3}")

# Solution: Use math.isclose for comparisons
import math
print(f"Is it close enough? {math.isclose(val, 0.3)}")

# For money/financials, use Decimal
from decimal import Decimal
d1 = Decimal('0.1')
d2 = Decimal('0.2')
print(f"Decimal sum: {d1 + d2}")
```
