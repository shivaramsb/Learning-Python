## 📝 Section 1: Variables and Data Types

Learn about Python's basic data types and how to work with them.

```python
# Different data types
name = "Alice"          # String
age = 25               # Integer
height = 5.6           # Float
is_student = True      # Boolean

print(f"Name: {name} (type: {type(name).__name__})")
print(f"Age: {age} (type: {type(age).__name__})")
print(f"Height: {height} (type: {type(height).__name__})")
print(f"Is Student: {is_student} (type: {type(is_student).__name__})")
```

### Type Conversion

Converting between different data types is a common operation.

```python
# Type conversion examples
age_str = "30"
age_int = int(age_str)
print(f"'{age_str}' (string) converted to {age_int} (integer)")

price_int = 20
price_float = float(price_int)
print(f"{price_int} (integer) converted to {price_float} (float)")

num = 100
num_str = str(num)
print(f"{num} (integer) converted to '{num_str}' (string)")
```

## ➕ Section 2: Arithmetic Operators

Perform mathematical operations with Python.

```python
a = 17
b = 5

print(f"a = {a}, b = {b}")
print(f"\nArithmetic Operations:")
print(f"Addition: {a} + {b} = {a + b}")
print(f"Subtraction: {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division: {a} / {b} = {a / b}")
print(f"Floor Division: {a} // {b} = {a // b}")
print(f"Modulus: {a} % {b} = {a % b}")
print(f"Exponent: {a} ** 2 = {a ** 2}")
```

### Practical Example: Calculate Average

```python
# Calculate average of three grades
num1 = 85
num2 = 92
num3 = 78
average = (num1 + num2 + num3) / 3

print(f"Grades: {num1}, {num2}, {num3}")
print(f"Average: {average:.2f}")
```

---
## ⚖️ Section 3: Comparison Operators

Compare values and get True/False results.

```python
x = 10
y = 5

print(f"x = {x}, y = {y}")
print(f"\nComparison Results:")
print(f"x == y: {x == y}")
print(f"x != y: {x != y}")
print(f"x > y: {x > y}")
print(f"x < y: {x < y}")
print(f"x >= 10: {x >= 10}")
print(f"y <= 5: {y <= 5}")
```

---
## 🔗 Section 4: Logical Operators

Combine conditions with AND, OR, and NOT.

```python
age = 22
has_id = True
is_member = False

print(f"Age: {age}, Has ID: {has_id}, Is Member: {is_member}")
print(f"\nLogical Operations:")
print(f"Can enter (age >= 18 AND has_id): {(age >= 18) and has_id}")
print(f"Gets discount (age < 18 OR is_member): {(age < 18) or is_member}")
print(f"NOT is_member: {not is_member}")

# Combining logical operators
can_purchase = (age >= 21) and (has_id) and (not is_member)
print(f"Can purchase alcohol: {can_purchase}")
```

## 📝 Section 5: String Operations

Work with text data in Python.

```python
first_name = "John"
last_name = "Doe"

# Concatenation
full_name = first_name + " " + last_name
print(f"Full name: {full_name}")

# Repetition
separator = "-" * 30
print(separator)
```

### String Methods

```python
text = "  Python Programming  "

print(f"Original: '{text}'")
print(f"Lowercase: '{text.lower()}'")
print(f"Uppercase: '{text.upper()}'")
print(f"Stripped: '{text.strip()}'")
print(f"Length: {len(text)}")
```

## 🎨 Section 6: String Formatting (f-strings)

Modern way to format strings in Python.

```python
name = "Alice"
age = 28
salary = 75000.50

# Basic f-string
print(f"Employee: {name}, Age: {age}")

# Expressions in f-strings
print(f"In 5 years, {name} will be {age + 5} years old")

# Formatting numbers
print(f"Salary: ${salary:,.2f}")  # Add comma separator and 2 decimals
```

### Table Formatting

```python
# Alignment in f-strings
print(f"{'Name':<10} {'Age':<5} {'Salary':<12}")
print(f"{name:<10} {age:<5} ${salary:<11,.2f}")
```

## 🔧 Section 8: Practical Examples

### Example 1: Temperature Conversion

```python
celsius = 25
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C = {fahrenheit}°F")
```

### Example 2: Shopping Cart Calculator

```python
item_name = "Laptop"
price = 899.99
quantity = 2
tax_rate = 0.08

subtotal = price * quantity
tax = subtotal * tax_rate
total = subtotal + tax

print(f"--- Receipt ---")
print(f"Item: {item_name}")
print(f"Price: ${price:.2f}")
print(f"Quantity: {quantity}")
print(f"Subtotal: ${subtotal:.2f}")
print(f"Tax (8%): ${tax:.2f}")
print(f"Total: ${total:.2f}")
print(f"---------------")
```

### Example 3: Circle Calculations

```python
radius = 7
pi = 3.14159

area = pi * (radius ** 2)
circumference = 2 * pi * radius

print(f"Circle with radius {radius}:")
print(f"Area: {area:.2f}")
print(f"Circumference: {circumference:.2f}")
```

### Example 4: Data Type Checking

```python
values = [42, 3.14, "Hello", True, None]

print("Data type checking:")
for value in values:
    print(f"{value} is of type: {type(value).__name__}")
```

## ⚠️ Section 9: Common Mistakes to Avoid

### Mistake 1: Type Mismatch

```python
age_string = "25"

# This would cause an error:
# result = age_string + 5  # TypeError!

# Correct way:
result = int(age_string) + 5
print(f"Correct: {result}")
```

### Mistake 2: Division Operators

```python
print(f"In Python 3, 5/2 = {5/2} (float)")
print(f"Use floor division for integer result: 5//2 = {5//2}")
```

### Mistake 3: Variable Naming

```python
# Wrong: my-variable, 2nd_place, class
# Correct:
my_variable = "correct"
second_place = "correct"
class_name = "correct"  # 'class' is a reserved keyword

print("All variable names are valid!")
```
